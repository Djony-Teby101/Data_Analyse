```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
np.random.seed(42)
n_transactions=200

# Génération des données brutes
dates = [datetime(2024, 1, 1) + timedelta(days=np.random.randint(0, 90)) for _ in range(n_transactions)]

data = {
    'transaction_id': list(range(1001, 1001 + n_transactions)),
    'date': dates,
    'montant': np.random.uniform(-5000, 10000, n_transactions).round(2),
    'categorie': np.random.choice(['Alimentation', 'Transport', 'Loisirs', 'Salaire', '???', '', None], n_transactions),
    'description': [f"ACHAT {np.random.choice(['CARREFOUR', 'UBER', 'NETFLIX', 'SALAIRIE', ''])}" for _ in range(n_transactions)],
    'type_client': np.random.choice(['particulier', 'pro', 'particulier', 'particulier', None], n_transactions),
    'devise': np.random.choice(['EUR', 'EUR', 'EUR', 'USD', 'GBP', ''], n_transactions)
}

df = pd.DataFrame(data)
```

```python
df.info() #Types, Valeur non nulles.
df.describe() #Stats sur colonnes numeriques
df.isnull().sum() #Comptage des valeur null.
```

```python
df.duplicated().sum()
df.drop_duplicates()
df.drop_duplicates(subset=["categorie"]) # Sur colonne specifiques
```

```python
# Remplace toutes les valeurs vides du DataFrame par 0
df["montant"]=df["categorie"].fillna("vide")
```
