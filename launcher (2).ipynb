# Nettoyage et Agregation des Données

```python
import pandas as pd
import numpy as np

# Lire le fichier.
df=pd.read_csv('ventes.txt')
print("lire le DataFrame.")
df.head(2)
```

```python
# Remplacer les valeur manquante par la moyenne des ventes de l'an 2021
moyenne_2021=df[df["Annee"]==2021]['Chiffre_Affaires'].mean().round(2)
print(f"la moyenne des ventes de l'année 2021 est {moyenne_2021}")
```

```python
# Agregation: Chiffre d'affaire Total par Ville.
df.groupby("Ville")["Chiffre_Affaires"].sum()
```
